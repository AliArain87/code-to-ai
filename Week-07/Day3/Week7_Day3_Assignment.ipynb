## Step 5 — Deploy to Streamlit Community Cloud

1. Go to **share.streamlit.io**, sign in with GitHub.
2. **Create app → Deploy a public app from GitHub.**
3. Pick your repo, branch `main`, main file `app.py`. Click **Deploy**.
4. Wait for the build. Copy your public link.
5. Paste the live link into your `README.md` and push again.

If the build errors, re-read the error line: it is almost always a missing
library in `requirements.txt` or a misnamed data file. Fix, push, and it
redeploys automatically.
