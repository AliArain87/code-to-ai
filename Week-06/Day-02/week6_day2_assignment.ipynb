## Setup — run this once

This cell builds your messy dataset. You do not need to understand every line of it —
it is just a machine for creating mess.

````python
import pandas as pd
import seaborn as sns

# Take a small, reproducible sample of the real diamonds data
raw = sns.load_dataset('diamonds').sample(n=200, random_state=42).reset_index(drop=True)

dirty = pd.DataFrame({
    'Diamond Cut':   raw['cut'].astype(str),
    'Color Grade':   raw['color'].astype(str),
    'Clarity Level': raw['clarity'].astype(str),
    'Carat Weight':  raw['carat'].astype(str),
    'Price USD':     raw['price'].astype(str),
    'Table Pct':     raw['table']
})

# Inject messy capitalisation and stray spaces into the text columns
def messify(series, i):
    out = []
    for n, value in enumerate(series):
        if (n + i) % 3 == 0:
            out.append('  ' + value.lower())
        elif (n + i) % 3 == 1:
            out.append(value.upper() + ' ')
        else:
            out.append(value.title())
    return out

dirty['Diamond Cut'] = messify(dirty['Diamond Cut'], 0)
dirty['Color Grade'] = messify(dirty['Color Grade'], 1)
dirty['Clarity Level'] = messify(dirty['Clarity Level'], 2)

# Break some prices so the column becomes text
dirty.loc[5, 'Price USD'] = 'not recorded'
dirty.loc[37, 'Price USD'] = 'N/A'
dirty.loc[88, 'Price USD'] = 'unknown'

# Add a fake purchase date stored as text
dirty['Purchase Date'] = pd.date_range('2026-01-01', periods=200, freq='D').astype(str)

# Duplicate 15 rows and shuffle them in
dupes = dirty.sample(n=15, random_state=7)
dirty = pd.concat([dirty, dupes], ignore_index=True)
dirty = dirty.sample(frac=1, random_state=3).reset_index(drop=True)

print("Your messy dataset is ready.")
print("Shape:", dirty.shape)
dirty.head(10)
````
